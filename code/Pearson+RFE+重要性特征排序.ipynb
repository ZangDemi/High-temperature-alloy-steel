```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib
import numpy as np
```

```python
# 读取数据
file_path = r'D:\creep rupture\添加高温测试和高应力\pearson+RFE+特征重要性排序\TF-IDF1.xlsx'
data = pd.read_excel(file_path)

# 计算皮尔逊相关系数矩阵
correlation_matrix = data.corr(method='pearson')

# 设置阈值
threshold = 0.85

# 创建一个布尔掩码，只保留大于阈值的相关系数
mask = np.abs(correlation_matrix) > threshold

# 设置全局字体为 Times New Roman
plt.rcParams['font.family'] = 'Times New Roman'

# 创建只包含大于阈值相关系数的注释数组，并将其他值设为空字符串
annot = correlation_matrix.where(mask).round(2).astype(str)
annot = annot.where(mask, "")

# 替换cm^3等单位符号
annot = annot.applymap(lambda x: x.replace('cm^3', 'cm³') if isinstance(x, str) else x)

# 创建 figsize
plt.figure(figsize=(12, 10))

# 设置 heatmap 参数
heatmap = sns.heatmap(
    correlation_matrix, 
    cmap='coolwarm', 
    vmin=-1, vmax=1, 
    linewidths=0.5, linecolor='gray', 
    cbar=True, annot=annot, fmt="", 
    annot_kws={"fontsize": 8, "fontweight": 'bold', "fontfamily": "Arial Unicode MS"}
)

# 设置标题和标签字体为 Times New Roman，加粗
plt.title('Feature Correlation Heatmap (Threshold > 0.85)', fontsize=18, fontweight='bold', fontname='Times New Roman')
plt.xticks(fontsize=12, fontweight='bold', fontname='Times New Roman')
plt.yticks(fontsize=12, fontweight='bold', fontname='Times New Roman')

# 调整边缘并保存为PDF和PNG文件
plt.tight_layout()
plt.savefig(r'D:\creep rupture\添加高温测试和高应力\pearson+RFE+特征重要性排序\0.85filtered_heatmap.pdf', format='pdf')
plt.savefig(r'D:\creep rupture\添加高温测试和高应力\pearson+RFE+特征重要性排序\0.85filtered_heatmap.png', format='png')
plt.show()

# 将大于阈值的相关系数对保存到Excel文件中
high_corr_pairs = correlation_matrix[mask].stack().reset_index()
high_corr_pairs = high_corr_pairs[high_corr_pairs['level_0'] != high_corr_pairs['level_1']]  # 去除自相关
high_corr_pairs.columns = ['Feature 1', 'Feature 2', 'Correlation']

# 保存到Excel文件
output_excel_path = r'D:\creep rupture\添加高温测试和高应力\pearson+RFE+特征重要性排序\0.85high_correlation_pairs.xlsx'
high_corr_pairs.to_excel(output_excel_path, index=False)

print(f"Filtered correlation pairs saved to:\n{output_excel_path}")
```

```python
# 必须导入的库
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor as RFR
from sklearn.feature_selection import RFE
import matplotlib.pyplot as plt
import numpy as np
from sklearn.model_selection import cross_val_score
```

```python
# Load your Excel file
file_path = r'D:\creep rupture\添加高温测试和高应力\pearson+RFE+特征重要性排序\0.85筛选新.xlsx'
data = pd.read_excel(file_path)

# Splitting the dataset into train and test sets (9:1 ratio)
train_data, test_data = train_test_split(data, test_size=0.1, random_state=42)

# Define file paths for saving
train_file_path = r'D:\creep rupture\添加高温测试和高应力\pearson+RFE+特征重要性排序\0.85train_data新.xlsx'
test_file_path = r'D:\creep rupture\添加高温测试和高应力\pearson+RFE+特征重要性排序\0.85test_data新.xlsx'

# Save train and test sets to Excel files
train_data.to_excel(train_file_path, index=False)
test_data.to_excel(test_file_path, index=False)

print("Train and test datasets saved successfully.")
```

```python
data=pd.read_excel(r'D:\creep rupture\添加高温测试和高应力\pearson+RFE+特征重要性排序\0.85train_data新.xlsx')
data.dropna(subset=['Tensile Strength(Mpa)'], inplace=True)  # 删除空缺值

X = data.drop(columns=['Tensile Strength(Mpa)', 'Time to Rupture(h)'])
y = data['Tensile Strength(Mpa)']
```

```python
estimator = RFR()

print("递归特征消除法和曲线图选取最优特征数量")

score = []
num_features = len(X.columns)
for i in range(1, num_features + 1):
    selector = RFE(estimator, n_features_to_select=i, step=1)
    selector = selector.fit(X, y)
    print(selector.support_)
    print(selector.ranking_)
    print(selector.n_features_)
    once = cross_val_score(estimator, selector.transform(X), y, cv=10).mean()
    score.append(once)

print("最佳验证得分:", max(score), "对应特征数量:", score.index(max(score)) + 1)
print(score)
```

```python
def objective_func(particle_position):
    max_depth = int(particle_position[0])
    learning_rate = particle_position[1]
    n_estimators = int(particle_position[2])
    min_child_weight = int(particle_position[3])
    subsample = particle_position[4]
    gamma = particle_position[5]

    model = xgb.XGBRegressor(max_depth=max_depth,
                             learning_rate=learning_rate,
                             n_estimators=n_estimators,
                             min_child_weight=min_child_weight,
                             subsample=subsample,
                             gamma=gamma)
    model.fit(x_train, y_train)
    y_pred = model.predict(x_test)
    mse = mean_squared_error(y_test, y_pred)
    return mse
```

```python
# 分割训练集的特征和目标变量
X_train = train_data[features]
y_train = train_data['Tensile Strength(Mpa)']

# 分割测试集的特征和目标变量
X_test = test_data[features]
y_test = test_data['Tensile Strength(Mpa)']

# 特征缩放
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 训练随机森林回归模型
rf = RandomForestRegressor(n_estimators=100, random_state=42)
rf.fit(X_train_scaled, y_train)

# 提取特征重要性
importances = rf.feature_importances_
feature_importances = pd.DataFrame({'Feature': features, 'Importance': importances})
feature_importances = feature_importances.sort_values(by='Importance', ascending=False)
```
